### 2.2 DATA PREPARATION 
____

```python
file = r'/Users/teslim/OneDrive/generaldata/11_world_population_data.csv'

df = pd.read_csv(file)
df.head()
```

```python
# Check the data types of the columns
df.info()
```

```python
# Droping some column that are not needed
# since the inplace = True.. needless to re-assigned the df again

df.drop(['CCA3', 'Rank'], axis=1, inplace=True)
```

```python
# confirm if the 'CCA3' column is drop.
df
```

```python
# Rename some column to make them more descriptive
df.rename(columns={ 'Country/Territory': 'Country',
                    'World Population Percentage':'population',
                    'Area (km²)':'Area',
                    'Density (per km²)': 'Density'}, inplace=True)

# check the dataset
df
```

```python
# change the column names to lower case, and use the .str.replace() to replace spaces with underscores
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.columns
```

```python
# checking the content of the dataset
df.head()
```

```python
# çhecking the data types of the columns. We can use the .dtypes attribute.
# object means string (categories or text),
# int64 means integer, float64 means float (decimal numbers

df.dtypes
```

```python
# Set data which are object(categorica) to indicates trues
df.dtypes == 'object'
```

```python
# use the select_dtypes to identify the categoterical and assign it to a variable
categorical_columns = df.select_dtypes(include='object')
categorical_columns
```

```python
# replace the values in the categorical columns with the lowercase values, and remove the spaces
for col in categorical_columns:
    df[col] = df[col].str.lower().str.replace(' ', '_')

df.head()
```

```python
# checking the unique count in each columns
df.nunique()
```

### 2.4 SETTING UP THE VALIDATION FRAMEWORK

```python
# checking the count of the data in the dataset
len(df)
```

```python
# We will need to confirm what is the 20% of the dataset, and turn it to an integer
len(df) * 0.2 # 20% of the dataset
int(len(df) * 0.2) # 20% of the dataset with the interger value
```

```python
# sliptting the data into 60%, 20%, and 20%.

n = len(df) # total number of samples
n_val = int(n * 0.2) # 20% of the dataset
n_test = int(n * 0.2) # 20% of the dataset
n_train = n - (n_val) - (n_test) # 60% of the dataset


# We will print the values of the n_train, n_val, n_test
print('The validation dataset total:', n_val)
print('The test dataset total:', n_test)
print('The train dataset total:', n_train)

print('Total', n_train + n_val + n_test)
```

```python
# The rows containing the df_val dataset 0 - 143

df_val = df.iloc[:n_val]
print(df_val.shape)
df_val
```

```python
# The rows containing the test dataset 46 - 92

df_test = df.iloc[n_val : n_val + n_test]
print(df_test.shape)
df_test
```

```python
# The rows containing the validation dataset 143 - 189

df_train = df.iloc[n_val + n_test: ]
print(df_train.shape)
df_train
```

```python
# Turning the dataset (n) into nupy arrays, and assigning the values to the variables

idx = np.arange(n)
idx
```

```python
# Shuffling the data, and setting the random seed to 2
np.random.seed(2)
np.random.shuffle(idx)
idx
```

```python
# After the shuffling, we will apply the shuffling data set to the train, val and test data

df_train_s = df.iloc[idx[:n_train]]
df_val_s= df.iloc[idx[n_train:n_train+n_val]]
df_test_s = df.iloc[idx[n_train+n_val:]]
```
